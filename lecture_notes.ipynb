# Pandas Data Structures
 - Series - Single dimensional data structure. Similar to a list or a single column
   - homogenous (of the same data type)
 - DataFrame - multidimensional table
   - heterogenous (of many datatypes)

```python
ser1 = pd.Series([10, 20,30,40], index=['A', 'B', 'C', 'D'])
print(ser1)
```

```python
# Creating a series from a preexisting dictionary

d = {
    'b': 1,
    'a': 0,
    'c': 2,
    'd':10
}

ser2 = pd.Series(d)
```

```python
d1 = {
    'a':0.0,
    'b':1.0,
    'c':2.0
}

ser3 = pd.Series(d1)

print(ser3)
```

```python
calories = {
    'day1': 700,
    'day2': 2500,
    'day3': 1200
}

# Use indexes to choose certain values for a series
ser4 = pd.Series(calories, index = ['day1', 'day2'])

print(ser4)
```

```python
# If provided a scalar values, value will repeat to fill all index labels
ser5 = pd.Series(5.0, index=['a', 'b', 'c'])
print(ser5)
```

### Pandas Date & Time 
-   `pd.to_datetime()`- to convert different data types into datetime object
-   `pd.date_range()`- used to generate a range of date and time 
-   `pd.timestamp()` - generate the time expressed as number of seconds since 01/01/1970

```python
# Generate a range of dates from start to end default frequency
# periods = the number of dates you want between the start and end (inclusive)
date_range = pd.date_range(start='04/05/2023', end='05/05/2026', periods=5)
print(date_range)
```

```python
date_range = pd.date_range(start='01/01/2020', end='05/05/2026', freq='QE')
date_range
```

```python
date_range = pd.date_range(start='01/01/2024', periods=6, tz='Asia/Tokyo')
date_range
```

```python
start_date = '01/01/2023'
num_days = 30
date_range = pd.date_range(start=start_date, periods=num_days, freq='D')

np.random.seed(42) # seed means we are all using the Same random algorithm

stock_prices = np.random.normal(loc=100, scale=5, size=num_days).round(2)

stock_df = pd.DataFrame({'Date': date_range, 'StockPrice': stock_prices})

stock_df
```

```python
url='https://raw.githubusercontent.com/bprasad26/lwd/master/data/tesla_stock_prices.csv'
# using parse_dates to change type from string into date
tesla_df = pd.read_csv(url)


# tesla_df.to_csv('data/tesla_stocks.csv')

tesla_df['Date'] = pd.to_datetime(tesla_df['Date']) 

tesla_df.head()


```

```python
# Add new columns to dataframe based off of dates



tesla_df['Day'] = tesla_df['Date'].dt.day
tesla_df['Month'] = tesla_df['Date'].dt.month
tesla_df['Year'] = tesla_df['Date'].dt.year
tesla_df['day_name'] = tesla_df['Date'].dt.day_name()

tesla_df.head(10)
```

```python
least_recent = tesla_df['Date'].min()
most_recent = tesla_df['Date'].max()

print(least_recent)
print(most_recent)
```

```python
monthly_sales = tesla_df.groupby(['Year', 'Month'])['Volume'].sum().reset_index()

monthly_sales.head(15)
```

### Filter Using loc[] or query()

```python
# filtered_df_date = tesla_df.loc[(tesla_df['Date'] >= '2010-08-01')]
filtered_df_date = tesla_df.loc[(tesla_df['Date'] >= '2010-08-01') & (tesla_df['Date'] <= '2010-12-31')]
 # created a df of a range of dates using conditions
 # conditional filtering
filtered_df_date
```

```python
filtered_df = tesla_df.query("Date >= '2010-08-01' and Date <= '2010-12-31'")
filtered_df
```
